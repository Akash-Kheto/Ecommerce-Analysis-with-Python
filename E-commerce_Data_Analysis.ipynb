```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Superstore.csv", encoding='latin-1') 
#encoding='latin-1': “Encoding is used to properly read and handle special characters.”
data.head()
```

# Convert  String to Date Columns

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date'] = pd.to_datetime(data['Ship Date']) 
```

```python
data.info()
```

```python
data.head()
```

# Adding New Date-Based Columns

```python
data['Order Month'] = data['Order Date'].dt.month 
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
```

```python
data.head()
```

# Monthly Sales Analysis

```python
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index()
sales_by_month
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Monthly sales calculate
sales_by_month = data.groupby("Order Month")["Sales"].sum().reset_index()

# Plot
plt.figure(figsize=(10, 6))

sns.lineplot(
    data=sales_by_month,
    x="Order Month",
    y="Sales",
    marker="o"
)
plt.title("Monthly Sales Analysis")
plt.xlabel("Order Month")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13))
plt.grid(True)

plt.show()
```

# Sales Analysis by Category

```python
sales_by_category = data.groupby("Category")["Sales"].sum().reset_index()
sales_by_category
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    sales_by_category["Sales"],
    labels=sales_by_category["Category"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.5)
)


plt.title(
    "Sales Analysis by Category",
    fontsize=24
)

plt.show()
```

# Sales Analysis by Sub-Category

```python
sales_by_subcategory = data.groupby("Sub-Category")["Sales"].sum().reset_index()
sales_by_category
```

```python
plt.figure(figsize=(12, 6))  # width = 12  , height = 6

sns.barplot(
    data=sales_by_subcategory,
    x="Sub-Category",
    y="Sales"
)

# Chart title and labels
plt.title("Sales Analysis by Sub-Category", fontsize=20)
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")

# Rotate x-axis labels
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# Monthly Profit Analysis

```python
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors

profit_by_month = data.groupby("Order Month")["Profit"].sum().reset_index()

fig = px.line(
    profit_by_month,
    x="Order Month",
    y="Profit",
    title="Monthly Profit Analysis"
)

fig.show()
```

# Profit Analysis by Category

```python
profit_by_category = data.groupby('Category')['Profit'].sum().reset_index()
profit_by_category
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    profit_by_category["Profit"],
    labels=profit_by_category["Category"],
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.5)
)

plt.title(
    "Profit Analysis by Category",
    fontsize=24
)

plt.tight_layout()
plt.show()
```

# profit_by_subcategory

```python
profit_by_subcategory = data.groupby('Sub-Category')['Profit'].sum().reset_index()
fig = px.bar(profit_by_subcategory, x='Sub-Category', 
             y='Profit', 
             title='Profit Analysis by Sub-Category')
fig.show()
```

# Sales and Profit Analysis by Customer Segment

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()

color_palette = colors.qualitative.Pastel

fig = go.Figure()
fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'], 
                     y=sales_profit_by_segment['Sales'], 
                     name='Sales',
                     marker_color=color_palette[0]))

fig.add_trace(go.Bar(x=sales_profit_by_segment['Segment'], 
                     y=sales_profit_by_segment['Profit'], 
                     name='Profit',
                     marker_color=color_palette[1]))

fig.update_layout(title='Sales and Profit Analysis by Customer Segment',
                  xaxis_title='Customer Segment', yaxis_title='Amount')

fig.show()
```

#   Analyse sales-to-profit ratio

```python
sales_profit_by_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```
